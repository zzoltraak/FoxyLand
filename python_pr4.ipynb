**Завдання 1**

Напишіть програму, яка отримує два рядки, і формує з них словник. Ключами є слова з першого рядка, значеннями - цілі числа з другого.



Виконайте приклад нижче

```python
# Ввід двох рядків
keys_line = input("Введіть слова (через пробіл): ")
values_line = input("Введіть числа (через пробіл): ")

# Розбиття рядків на списки
keys = keys_line.split()
values = list(map(int, values_line.split()))

# Формування словника
result = dict(zip(keys, values))

# Вивід результату
print(result)

```

**Завдання 2**

Напишіть програму, яка отримує на вхід рядок та підраховує, скільки разів в ньому зустрічається кожен символ (незалежно від регістру). Результат слід вивести без фігурних дужок.


Виконайте приклад нижче:

```python
text = input("Введіть текст: ").lower()

# Створення словника через генератор
char_count = {char: text.count(char) for char in set(text)}

# Вивід без фігурних дужок
for char, count in char_count.items():
    print(f"{char}: {count}")
```

**Завдання 4**

Напишіть програму, яка формує телефонний довідник із введеного користувачем рядка, а потім видає потрібний номер на ім'я абонента. Якщо абонента з такою назвою немає, потрібно вивести Немає такого абонента.

Виконайте приклад нижче:

```python
# Ввід телефонного довідника у форматі: ім'я номер
data = input("Введіть довідник (наприклад: Олег 12345 Марія 67890): ")

# Розбиття рядка на список
parts = data.split()

# Формування словника: кожне парне слово — ім’я, непарне — номер
phonebook = {parts[i]: parts[i+1] for i in range(0, len(parts), 2)}

# Ввід імені для пошуку
name = input("Введіть ім'я абонента: ")

# Вивід результату
if name in phonebook:
    print(phonebook[name])
else:
    print("Немає такого абонента")
```

**Завдання для самостійного виконання 4**

Додайте до цього словника ще 2 пари номерів та абонентів. Перевірте, що немає однакових номерів

```python
data = input("Введіть довідник (наприклад: Олег 12345 Марія 67890): ")

parts = data.split()
phonebook = {parts[i]: parts[i + 1] for i in range(0, len(parts), 2)}

for i in range(2):
    name = input("Введіть ім'я нового абонента: ")
    number = input("Введіть номер нового абонента: ")

    if number in phonebook.values():
        print("Такий номер вже існує")
    else:
        phonebook[name] = number

print(phonebook)
```

**Завдання 6**

Напишіть програму, яка:
Приймає на вхід 3 рядки з назвами країн та розташованими в них містами.
Отримує від користувача назву міста та виводить країну, в якій він знаходиться.


Виконайте приклад нижче:


```python
# Ввід трьох рядків з країнами та містами
country1 = input("Перша країна та її міста: ")  # Наприклад: Україна Київ Львів Харків
country2 = input("Друга країна та її міста: ")  # Наприклад: Франція Париж Ліон Ніцца
country3 = input("Третя країна та її міста: ")  # Наприклад: Італія Рим Мілан Турин

# Формування словника: місто → країна
city_to_country = {}

for line in [country1, country2, country3]:
    parts = line.split()
    country = parts[0]
    cities = parts[1:]
    for city in cities:
        city_to_country[city] = country

# Ввід назви міста для пошуку
search_city = input("Введіть назву міста: ")

# Вивід результату
if search_city in city_to_country:
    print(city_to_country[search_city])
else:
    print("Немає такого міста в довіднику")
```

Створює словник, де ключ — назва країни, значення — столиця

Отримує від користувача назву країни

Виводить назву столиці цієї країни

Якщо країни немає в довіднику — виводить Немає такої країни

```python
line1 = input("Введіть країну та столицю: ")
line2 = input("Введіть країну та столицю: ")
line3 = input("Введіть країну та столицю: ")

capitals = {}

for line in [line1, line2, line3]:
    parts = line.split()
    country = parts[0]
    capital = parts[1]
    capitals[country] = capital

search_country = input("Введіть назву країни: ")

if search_country in capitals:
    print(capitals[search_country])
else:
    print("Немає такої країни")
```

**Завдання для самостійного виконання 9**

Напишіть програму, яка:

Отримує список слів (через пробіл)

Створює словник, де ключ — слово, а значення — кількість його входжень

Виводить словник без фігурних дужок

```python
words = input("Введіть слова через пробіл: ").split()

word_count = {}

for word in words:
    word_count[word] = word_count.get(word, 0) + 1

for word, count in word_count.items():
    print(f"{word}: {count}")
```
